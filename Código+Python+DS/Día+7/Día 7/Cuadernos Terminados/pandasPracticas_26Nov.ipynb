#### Crear un DataFrame: cuatro caminos

```python
# 1. Diccionario de listas. Cada clave es una columna; cada lista, sus valores. Pandas convierte el diccionario en una tabla y asigna un índice 0, 1, 2
Datos = {'nombre':['Pedro','Juan','Lorena'],'edad':[25,39,33]}
df = pd.DataFrame(Datos)
df
```

```python
Datos = [['Pedro','Juan','Lorena'],[25,39,33]]
df = pd.DataFrame(Datos)
df
```

```python
list_dict = [{'a':1,'b':4,'c':7},{'a':2,'b':5,'c':8}]
df_dict = pd.DataFrame(list_dict)
df_dict
```

```python
data = [['Amy White', 50, 5], ['Jack Stewart', 53, 8]]
df = pd.DataFrame(data, columns=['Name', 'Age', 'Working Experience (Yrs.)'])
df
```

```python
array_a = np.array([[1000,50,78.45],[700,112,35.5]])
print(type(array_a))
```

```python
array_a = np.array([[1000,50,78.45],[700,112,35.5]])
company_statistics = pd.DataFrame(array_a,
    columns=['No. of Employees', 'Avg. Annual Revenue ($)', 'Share price ($)'],
    index=['Company A', 'Company B'])
company_statistics
```

```python
company_statistics = company_statistics.reset_index()
company_statistics
```

```python
company_statistics = company_statistics.reset_index(drop=True)
company_statistics
```
