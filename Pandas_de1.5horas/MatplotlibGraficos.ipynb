### MatpotLib Seccion de Graficos 

```python
## importamos la libreria MatPLotLib
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
## Datos de ejemplo
Ventas = pd.read_csv(r"C:\Users\rfigu\Documents\Learn\CARRERA_DATOS\Pandas_de1.5horas\BasesEjemplo\VentasPorProveedor.csv" ## TablaEjemplo.csv"
                    , sep=";"
                    ,skiprows=5)
Ventas.head()
```

```python
ali = Ventas[Ventas.Categoría == 'Comestibles']
ali.head()
```

```python
ali.Artículo.value_counts().plot(kind='line')
```

```python
## el punto y coma elimina el titulo en el encabezado
ali.Artículo.value_counts().plot(kind='bar');
```

```python
ali.Artículo.value_counts().plot(kind='barh');
```

```python
ali.Artículo.value_counts().plot(kind='pie');
```

```python
ali.Artículo.value_counts().plot(kind='scatter');
```
