### Los 10 Errores Más Comunes en PYTHON

### SyntaxError: invalid syntax
¿Por qué ocurre? Python encontró algo que no entiende. Generalmente es un paréntesis, comilla o dos puntos que falta.

```python
if edad >= 18   # ← Falta el : al final
    print("Eres mayor de edad")
```

### IndentationError
¿Por qué ocurre? Python usa la indentación (espacios al inicio) para saber qué código va dentro de qué. Si la indentación está mal, Python se confunde.

S: Agrega 4 espacios (o 1 Tab) al inicio de las líneas que van dentro de un bloque. Asegúrate de no mezclar espacios y tabs en el mismo archivo.

```python
for i in range(5):
print(i)  # ← Falta indentación
```

### SyntaxError: EOL while scanning string

¿Por qué ocurre? Abriste una comilla pero nunca la cerraste. Python llegó al final de la línea (EOL = End Of Line) buscando la comilla de cierre.

```python
mensaje = "Hola, bienvenido al curso 
```

### NameError: name 'x' is not defined
¿Por qué ocurre? Estás usando una variable que no existe. O la escribiste mal, o nunca la creaste.

```python
nombre = "Juan"
print(nomber)
```

### UnboundLocalError

¿Por qué ocurre? Dentro de una función, intentas usar una variable que Python cree que es local, pero todavía no tiene valor.
Solución
Si necesitas modificar una variable global dentro de una función, declara global contador al inicio de la función. O mejor: pasa la variable como parámetro.

```python
contador = 0

def incrementar():
    contador = contador + 1  # ← Error aquí
    return contador
```

```python
incrementar()
```
